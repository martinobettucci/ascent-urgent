## 2. Architecture
LSTM reçoit (lot, temps, variable) avec batch_first=True. La tête utilise le dernier état produit pour prédire une valeur. Le clipping dans la boucle limite la norme des gradients ; il n’élimine pas tous les problèmes d’optimisation.

```python
class MonLSTM(nn.Module):
    def __init__(self):
        super().__init__();self.lstm=nn.LSTM(1,16,batch_first=True);self.sortie=nn.Linear(16,1)
    def forward(self,x):
        etats,_=self.lstm(x)
        return self.sortie(etats[:,-1,:]).squeeze(-1)
seed_all();m=MonLSTM();assert tuple(m(torch.tensor(d['train'][0][:5])).shape)==(5,)
h=fit_model(m,d['train'],d['validation'],task='regression',epochs=15,lr=.003)
plot_history(h,'Prévision LSTM','06_lstm.png')
```

## 3. Revenir à l’unité originale
La perte d’entraînement porte sur des valeurs standardisées. Le métier doit recevoir une MAE/RMSE dans les unités initiales. La baseline de persistance prévoit la dernière observation connue.

```python
X,y=d['validation'];vrai=y*d['sd']+d['mu']
pred=predict(m,X,'regression')*d['sd']+d['mu'];naive=X[:,-1,0]*d['sd']+d['mu']
mesures={name:{'MAE':float(mean_absolute_error(vrai,values)),'RMSE':float(mean_squared_error(vrai,values)**.5)} for name,values in [('LSTM',pred),('Persistance',naive)]}
print(pd.DataFrame(mesures).T)
fig,ax=plt.subplots(figsize=(8,4));ax.plot(vrai[:96],label='Observé');ax.plot(pred[:96],label='LSTM');ax.plot(naive[:96],label='Persistance',linestyle=':')
ax.set(xlabel='Pas de validation',ylabel='Unité originale');ax.legend();fig.tight_layout();fig.savefig(RESULTS/'06_previsions.png',dpi=150)
save_result('06_sequence',{'validation':mesures,'test_ouvert':False,'fenetre':24})
```
